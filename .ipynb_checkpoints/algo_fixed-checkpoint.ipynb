# Cultural Algorithm for Wordle Solver
## An Innovative Approach to Word Guessing Games

This notebook implements a Cultural Algorithm to solve Wordle puzzles by:
1. Maintaining a population of candidate word guesses
2. Using cultural knowledge (letter position frequencies) to guide evolution
3. Combining genetic operators with belief-space knowledge

```python
import random
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict, Counter
```

```python
# Load word list
with open("words.txt", "r") as f:
    words = [line.strip() for line in f]

print(f"Loaded {len(words)} words")
```

## Wordle Game Mechanics
These functions simulate Wordle's color feedback system

```python
def grey(guessed_word, correct_word, pos):
    """Check if letter at position is grey (not in word)"""
    if guessed_word[pos] in (set(guessed_word) - set(correct_word)):
        return True
    else:
        return False

def green(guessed_word, correct_word, pos):
    """Check if letter at position is green (correct position)"""
    if guessed_word[pos] == correct_word[pos]:
        return True
    else:
        return False

def yellow(guessed_word, correct_word, pos):
    """Check if letter at position is yellow (in word, wrong position)"""
    if grey(guessed_word, correct_word, pos) or green(guessed_word, correct_word, pos):
        return False
    else:
        return True

def color_for_letter(guessed_word, correct_word, pos):
    """Return color code for visualization"""
    if green(guessed_word, correct_word, pos):
        return "#6aaa64"  # Green
    elif yellow(guessed_word, correct_word, pos):
        return "#c9b458"  # Yellow
    else:
        return "#787c7e"  # Grey
```

## Fitness Function
Evaluates how good a guess is compared to the target word

```python
def fitness(guessed_word, correct_word):
    """Calculate fitness score for a guessed word
    
    Returns:
        1.0 if perfect match
        0.2 per green letter (correct position)
        0.1 per yellow letter (correct letter, wrong position)
    """
    score = 0
    if guessed_word == correct_word:
        score = 1
    else:
        # FIXED: Changed range(4) to range(5) to check all 5 letters
        for pos in range(5):  
            if green(guessed_word, correct_word, pos):
                score += 0.2
            elif yellow(guessed_word, correct_word, pos):
                score += 0.1
    return score
```

## Position Frequency Analysis
Calculate how often each letter appears at each position in the word list

```python
def position_frequencies(words):
    """Calculate frequency of each letter at each position (0-4)"""
    freq = defaultdict(lambda: [0]*5)

    for word in words:
        for i, ch in enumerate(word):
            freq[ch][i] += 1

    # Convert raw counts to frequencies
    total_words = len(words)
    for ch in freq:
        freq[ch] = [count / total_words for count in freq[ch]]

    return dict(freq)

def score_word(word, pos_freq):
    """Score a word based on position frequencies"""
    score = 0.0
    for i, ch in enumerate(word):
        score += pos_freq.get(ch, [0]*5)[i]
    return score

def score_all_words(words, pos_freq):
    """Score all words and return sorted by score"""
    scored = {word: score_word(word, pos_freq) for word in words}
    return dict(sorted(scored.items(), key=lambda x: x[1], reverse=True))
```

## Cultural Algorithm Components

```python
def initialize_population(pop_size, words):
    """Initialize population with random words from word list"""
    population = random.sample(words, min(pop_size, len(words)))
    return population
```

```python
def evaluate_population(population, correct_word):
    """Evaluate fitness for all individuals in population"""
    return [fitness(ind, correct_word) for ind in population]
```

```python
def accept_individuals(population, fitness_scores, acceptance_ratio):
    """Select top individuals to update belief space"""
    n_accept = int(len(population) * acceptance_ratio)
    indices = np.argsort(fitness_scores)[-n_accept:]  # Highest fitness
    return [population[i] for i in indices], [fitness_scores[i] for i in indices]
```

```python
def update_belief_space(belief_space, accepted_pop, accepted_fit, population, words):
    """Update cultural knowledge from accepted individuals"""
    # Calculate position frequencies for normative knowledge
    pos_freq = position_frequencies(words)
    item_frequency = score_all_words(population, pos_freq)

    # Store frequency information
    belief_space['normative']['item_frequency'] = item_frequency
    belief_space['normative']['total_accepted'] = len(accepted_pop)
    
    # Situational knowledge: best solution found
    best_idx = np.argmax(accepted_fit)
    if accepted_fit[best_idx] > belief_space['situational']['best_fitness']:
        belief_space['situational']['best_solution'] = accepted_pop[best_idx]
        belief_space['situational']['best_fitness'] = accepted_fit[best_idx]
```

```python
def evolve_population(population, fitness_scores, words, mutation_rate=0.3):
    """Apply genetic operators: selection, crossover, mutation"""
    pop_size = len(population)
    new_population = []
    
    # Elitism: keep best individual
    best_idx = np.argmax(fitness_scores)
    new_population.append(population[best_idx])
    
    while len(new_population) < pop_size:
        # Tournament selection
        idx1, idx2 = np.random.choice(pop_size, 2, replace=False)
        parent1 = population[idx1] if fitness_scores[idx1] > fitness_scores[idx2] else population[idx2]
        
        idx1, idx2 = np.random.choice(pop_size, 2, replace=False)
        parent2 = population[idx1] if fitness_scores[idx1] > fitness_scores[idx2] else population[idx2]
        
        # Crossover: combine letters from both parents
        crossover_point = random.randint(1, 4)
        offspring = parent1[:crossover_point] + parent2[crossover_point:]
        
        # Mutation: replace with random word from word list
        if np.random.random() < mutation_rate:
            offspring = random.choice(words)
        
        new_population.append(offspring)
    
    return new_population
```

## Main Cultural Algorithm

```python
def cultural_algorithm_wordle(correct_word, words, pop_size=50, acceptance_ratio=0.2, 
                               influence_ratio=0.3, generations=100):
    """
    Cultural Algorithm for Wordle solving.
    
    Args:
        correct_word: The target word to guess
        words: List of valid words
        pop_size: Population size
        acceptance_ratio: Fraction of population to accept into belief space
        influence_ratio: Fraction of population influenced by belief space
        generations: Number of generations to run
    
    Returns:
        best_solution: Best word found
        history: Fitness history over generations
    """
    
    print("Running Cultural Algorithm for Wordle")
    print(f"Target word: {'*' * len(correct_word)} (hidden)")
    print(f"Population: {pop_size}, Generations: {generations}")
    print("-" * 60)
    
    # Initialize population
    population = initialize_population(pop_size, words)
    fitness_scores = evaluate_population(population, correct_word)
    
    # Initialize belief space
    pos_freq = position_frequencies(words)
    initial_ranked = score_all_words(words, pos_freq)
    
    belief_space = {
        'normative': {
            'item_frequency': initial_ranked,
            'total_accepted': 0
        },
        'situational': {
            'best_solution': None,
            'best_fitness': float('-inf')
        }
    }
    
    history = []
    
    # Evolution loop
    for gen in range(generations):
        # 1. Accept: Select best individuals
        accepted_pop, accepted_fit = accept_individuals(
            population, fitness_scores, acceptance_ratio
        )
        
        # 2. Update belief space
        update_belief_space(belief_space, accepted_pop, accepted_fit, population, words)
        
        # 3. Evolve: Genetic operations
        population = evolve_population(population, fitness_scores, words)
        fitness_scores = evaluate_population(population, correct_word)
        
        # Track progress
        best_fitness = max(fitness_scores)
        history.append(best_fitness)
        
        if gen % 20 == 0:
            best_word = population[np.argmax(fitness_scores)]
            print(f"Generation {gen}: Best Fitness = {best_fitness:.2f}, Best Word = {best_word}")
        
        # Early stopping if solution found
        if best_fitness >= 1.0:
            print(f"\n✓ Solution found at generation {gen}!")
            break
    
    print("-" * 60)
    best_solution = belief_space['situational']['best_solution']
    best_fitness = belief_space['situational']['best_fitness']
    
    print(f"\nBest Solution: {best_solution}")
    print(f"Fitness Score: {best_fitness:.2f}")
    print(f"Correct Word: {correct_word}")
    print(f"Match: {best_solution == correct_word}")
    
    return best_solution, history
```

```python
def plot_convergence(history):
    """Visualize convergence over generations"""
    plt.figure(figsize=(10, 6))
    plt.plot(history, linewidth=2, color='#2E86AB')
    plt.xlabel('Generation', fontsize=12)
    plt.ylabel('Best Fitness', fontsize=12)
    plt.title('Cultural Algorithm - Wordle Solver Convergence', fontsize=14, fontweight='bold')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

## Run the Algorithm

```python
# Set the target word
correct_word = "queen"  # You can change this to any 5-letter word from your word list

# Run the algorithm
best_solution, history = cultural_algorithm_wordle(
    correct_word=correct_word,
    words=words,
    pop_size=50,
    acceptance_ratio=0.2,
    influence_ratio=0.3,
    generations=100
)

# Plot convergence
plot_convergence(history)
```

## Test with Multiple Words

```python
# Test with multiple random words
test_words = random.sample(words, 5)
results = []

for test_word in test_words:
    print(f"\n{'='*60}")
    print(f"Testing with word: {test_word}")
    print(f"{'='*60}")
    
    best_sol, hist = cultural_algorithm_wordle(
        correct_word=test_word,
        words=words,
        pop_size=50,
        acceptance_ratio=0.2,
        generations=100
    )
    
    results.append({
        'word': test_word,
        'found': best_sol == test_word,
        'generations': len(hist),
        'final_fitness': hist[-1] if hist else 0
    })

# Summary
print(f"\n\n{'='*60}")
print("SUMMARY")
print(f"{'='*60}")
success_rate = sum(1 for r in results if r['found']) / len(results) * 100
print(f"Success Rate: {success_rate:.1f}%")
print(f"Average Generations: {np.mean([r['generations'] for r in results]):.1f}")
for r in results:
    status = "✓" if r['found'] else "✗"
    print(f"{status} {r['word']}: Fitness={r['final_fitness']:.2f}, Gens={r['generations']}")
```
